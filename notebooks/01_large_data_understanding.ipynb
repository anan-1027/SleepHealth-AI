```python
import pandas as pd
import os

# Load the new 100,000 row dataset
file_path = os.path.join("..", "data", "raw", "large_sleep_data.csv")
df = pd.read_csv(file_path)

# Display basic info
print("Dataset Shape:", df.shape)
print("\nMissing Values:\n", df.isnull().sum())

# Display the first 3 rows directly in the notebook
display(df.head(3))
```
